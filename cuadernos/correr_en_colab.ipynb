```python
# 3. Parámetros de la corrida
ANIO = 2010
CONFIG   = f'/content/poblar_repo/aplicaciones/prais2026/configs/{ANIO}.yaml'
INSUMOS  = '/content/drive/MyDrive/PRAIS/poblar/insumos'      # tabla_censoAAAA.xlsx, radios_AAAA.zip, regiones_PRAIS.gpkg
SALIDA   = '/content/drive/MyDrive/PRAIS/poblar/corridas'
ETAPAS   = ['insumos', 'radios']
```
